# Map launch sites and recorded outcomes

The geographical extract includes 56 launches. Pan and zoom to inspect the recorded successes and other outcomes by launch site. Basemap tiles require a connection to the tile provider.

```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster

spacex_df = pd.read_csv('data/spacex_launch_geo.csv')
spacex_df.groupby('Launch Site')['class'].agg(['count','sum'])
```

```python
sites = spacex_df.groupby('Launch Site')[['Lat','Long']].first()
site_map = folium.Map(location=sites.mean().tolist(), zoom_start=4)
for name, site in sites.iterrows():
    folium.CircleMarker([site.Lat,site.Long], radius=8, tooltip=name).add_to(site_map)
cluster = MarkerCluster().add_to(site_map)
for _, row in spacex_df.iterrows():
    colour = 'green' if row['class'] == 1 else 'red'
    folium.Marker([row.Lat,row.Long], tooltip=f"{row['Date']} · {row['Landing Outcome']}",
                  icon=folium.Icon(color=colour)).add_to(cluster)
site_map
```

## Distance from a site to a selected coastline point

The coordinates below were read from the course map. The calculation is a point-to-point approximation.

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
launch_site = sites.loc['CCAFS LC-40']
reference = (28.56338, -80.56798)
distance = calculate_distance(launch_site.Lat, launch_site.Long, *reference)
print(f'Distance to the selected coastline point: {distance:.2f} km')
```
