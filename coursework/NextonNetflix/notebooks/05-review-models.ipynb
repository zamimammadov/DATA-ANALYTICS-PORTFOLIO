# Review sentiment on distinct labelled texts

Compare the original Multinomial Naive Bayes baseline and the application model using the same held-out reviews. Vocabulary is fitted within each training pipeline. The labelled examples come from the included reviews.csv.

```python
from pathlib import Path
import sys
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

project = Path.cwd()
if project.name == 'notebooks':
    project = project.parent
sys.path.insert(0, str(project))
from sentiment import labelled_reviews, fit_model

texts, labels = labelled_reviews()
train, test, train_labels, test_labels = train_test_split(
    texts, labels, test_size=0.2, stratify=labels, random_state=42
)
print(len(train), 'training reviews;', len(test), 'held out')
```

```python
baseline = make_pipeline(TfidfVectorizer(stop_words='english'), MultinomialNB())
baseline.fit(train, train_labels)
app_model = fit_model(train, train_labels)
for name, model in [('Naive Bayes', baseline), ('Logistic regression', app_model)]:
    print(name)
    print(classification_report(test_labels, model.predict(test), digits=3))
```
