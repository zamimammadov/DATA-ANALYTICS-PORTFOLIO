# Explore launch records with SQLite

The SQL exercise uses a separate 101-row course extract. The column names are normalized before loading the table. Dates and payload figures follow that extract.

```python
import sqlite3
import pandas as pd

con = sqlite3.connect(':memory:')
df = pd.read_csv('data/Spacex.csv').rename(columns={'Landing _Outcome':'Landing_Outcome'})
df.to_sql('SPACEXTABLE', con, index=False)
df.head()
```

## Tasks

Now write and execute SQL queries to solve the assignment tasks.

**Note: If the column names are in mixed case enclose it in double quotes
   For Example "Landing_Outcome"**

### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
pd.read_sql_query("""SELECT DISTINCT "Launch_Site"  
FROM SPACEXTABLE;""", con)
```


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA' 


```python
pd.read_sql_query("""SELECT * from SPACEXTABLE WHERE "Launch_Site" LIKE 'CCA%' LIMIT 5;""", con)
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
pd.read_sql_query("""SELECT SUM("Payload_Mass__kg_") AS Total_Payload_Mass 
FROM SPACEXTABLE  
WHERE "Customer" = 'NASA (CRS)';""", con)
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
pd.read_sql_query("""SELECT AVG("Payload_Mass__kg_") AS Avg_Payload_Mass
FROM SPACEXTABLE
WHERE "Booster_Version" LIKE 'F9 v1.1%';""", con)
```

### Task 5

##### List the date when the first succesful landing outcome in ground pad was acheived.


_Hint:Use min function_ 


```python
pd.read_sql_query("""SELECT MIN("Date") 
FROM SPACEXTABLE
WHERE "Landing_Outcome" LIKE 'Success%ground pad%';""", con)
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
pd.read_sql_query("""SELECT DISTINCT "Booster_Version" 
FROM SPACEXTABLE
WHERE "Landing_Outcome" LIKE 'Success%drone ship%'
  AND "Payload_Mass__kg_" > 4000
  AND "Payload_Mass__kg_" < 6000;""", con)
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
pd.read_sql_query("""SELECT "Mission_Outcome", COUNT(*) AS Outcome_Count
FROM SPACEXTABLE
GROUP BY "Mission_Outcome";""", con)
```

### Task 8



##### List all the booster_versions that have carried the maximum payload mass, using a subquery with a suitable aggregate function.


```python
pd.read_sql_query("""SELECT DISTINCT "Booster_Version", "Payload_Mass__kg_"
FROM SPACEXTABLE
WHERE "Payload_Mass__kg_" = (
    SELECT MAX("Payload_Mass__kg_")
    FROM SPACEXTABLE
);""", con)
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
pd.read_sql_query("""SELECT 
    substr("Date", 6, 2) AS Month,
    "Landing_Outcome",
    "Booster_Version",
    "Launch_Site"
FROM SPACEXTABLE
WHERE substr("Date", 0, 5) = '2015'
  AND "Landing_Outcome" LIKE 'Failure%drone ship%';""", con)
```

Rank landing outcomes for recorded launches between June 4, 2010 and March 20, 2017.

```python
pd.read_sql_query("""SELECT Landing_Outcome, COUNT(*) AS launches
FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY Landing_Outcome ORDER BY launches DESC""", con)
```
