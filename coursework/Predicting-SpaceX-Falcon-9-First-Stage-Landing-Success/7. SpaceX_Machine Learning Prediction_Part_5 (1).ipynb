# Evaluate Falcon 9 landing predictions in launch order

IBM Applied Data Science Capstone. This revision uses a chronological holdout and fits encoders and scaling inside each training fold.

```python
import pandas as pd
from evaluate import DATA, evaluate

launches = pd.read_csv(DATA)
launches.sort_values(['Date', 'FlightNumber']).groupby('Class').size()
```

## Select a classifier on training flights

The earliest 72 launches form the training set. Three expanding time splits compare the four model families by mean balanced accuracy. The last 18 flights are kept for evaluation.

```python
cv_scores, holdout = evaluate(launches)
cv_scores.round(3)
```

## Check the selected model against the majority baseline

```python
holdout.round(3)
```

The tree classified 16 of 18 holdout flights correctly, including one of the three class 0 flights. Predicting class 1 for every launch was correct 15 times. This small retrospective difference is sensitive to a single flight.
