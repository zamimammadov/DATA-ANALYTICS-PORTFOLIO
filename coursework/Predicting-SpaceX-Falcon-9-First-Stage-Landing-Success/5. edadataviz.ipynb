# Explore launch sites, payloads and orbital destinations

Plots use the IBM 90-flight labelled extract. Comparisons are descriptive and group sizes vary.

```python
# pandas is a software library written for the Python programming language for data manipulation and analysis.
import pandas as pd
#NumPy is a library for the Python programming language, adding support for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays
import numpy as np
# Matplotlib is a plotting library for python and pyplot gives us a MatLab like plotting framework. We will use this in our plotter function to plot data.
import matplotlib.pyplot as plt
#Seaborn is a Python data visualization library based on matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics
import seaborn as sns
```

```python
df = pd.read_csv('data/dataset_part_2.csv')
df.head()
```

```python
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect = 5)
plt.xlabel("Flight Number",fontsize=20)
plt.ylabel("Pay load Mass (kg)",fontsize=20)
plt.show()
```

```python
# Plot a scatter point chart with x axis to be Flight Number and y axis to be the launch site, and hue to be the class value


# catplot ile FlightNumber vs LaunchSite
sns.catplot(
    data=df, 
    x="FlightNumber", 
    y="LaunchSite", 
    hue="Class", 
    kind="strip",    # noktaları göstermek için
     
    aspect=3
)
plt.xlabel("Flight Number",fontsize=15)
plt.ylabel("Launch Site",fontsize=15)
plt.title("Flight Number vs Launch Site (by Class)")
plt.show()
```

```python
# Plot a scatter point chart with x axis to be Pay Load Mass (kg) and y axis to be the launch site, and hue to be the class value


sns.scatterplot(
    data=df, 
    x="PayloadMass", 
    y="LaunchSite", 
    hue="Class", 
    
)

plt.title("Payload Mass vs Launch Site (by Class)")
plt.xlabel("Payload Mass (kg)")
plt.ylabel("Launch Site")
plt.show()
```

```python
# HINT use groupby method on Orbit column and get the mean of Class column
import seaborn as sns
import matplotlib.pyplot as plt

# Grup bazında başarı oranı
orbit_success = df.groupby("Orbit")["Class"].mean().reset_index()

plt.figure(figsize=(12,6))
sns.barplot(data=orbit_success, x="Orbit", y="Class")
plt.title("Success Rate by Orbit Type")
plt.ylabel("Success Rate")
plt.xlabel("Orbit Type")
  # orbit isimlerini yan yatır ki okunaklı olsun
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(16,6))
sns.scatterplot(
    data=df,
    x="FlightNumber",
    y="Orbit",
    hue="Class",
  
)

plt.title("Flight Number vs Orbit (by Class)")
plt.xlabel("Flight Number")
plt.ylabel("Orbit Type")
plt.show()
```

```python
# Plot a scatter point chart with x axis to be Payload Mass and y axis to be the Orbit, and hue to be the class value


plt.figure(figsize=(16,6))
sns.scatterplot(
    data=df,
    x="PayloadMass",   # X ekseni → Payload
    y="Orbit",               # Y ekseni → Orbit (kategorik)
    hue="Class",             # Başarı durumu renklere göre
   
)

plt.title("Payload Mass vs Orbit (by Class)")
plt.xlabel("Payload Mass (kg)")
plt.ylabel("Orbit Type")
plt.show()
```

## Annual recorded landing rate

```python
annual = df.assign(year=pd.to_datetime(df['Date']).dt.year).groupby('year')['Class'].agg(['count', 'mean'])
annual['mean'].plot(marker='o', ylabel='Share of class 1 flights', ylim=(0, 1))
annual
```

## Explore categorical features

This table illustrates feature encoding. The evaluation notebook fits its own preprocessing on training data.

```python
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]
features.head()
```

```python
# HINT: Use get_dummies() function on the categorical columns

# Kategorik sütunları seç
categorical_cols = ["Orbit", "LaunchSite", "LandingPad", "Serial"]

# One-hot encoding uygula
features_one_hot = pd.get_dummies(features, columns=categorical_cols)

# İlk satırları göster
features_one_hot.head()
```
