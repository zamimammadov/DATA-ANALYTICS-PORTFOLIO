# Wrangle the Falcon 9 landing labels

The IBM course defines a binary target using the first token of Outcome. A successful ocean landing is class 1. Missing outcomes and flights with no attempted landing share class 0.

```python
import pandas as pd
df = pd.read_csv('data/dataset_part_1.csv')
df.head(5)
```

```python
df.isna().mean().mul(100)
```

```python
df['LaunchSite'].value_counts()
```

```python
df['Orbit'].value_counts()
```

```python
df['Outcome'].value_counts()
```

```python
df['Class'] = df['Outcome'].fillna('').str.startswith('True ').astype(int)
reference = pd.read_csv('data/dataset_part_2.csv')
pd.testing.assert_series_equal(df.set_index('FlightNumber')['Class'], reference.set_index('FlightNumber')['Class'])
df.groupby(['Outcome','Class']).size().to_frame('launches')
```

```python
df['Class'].value_counts().sort_index()
```
